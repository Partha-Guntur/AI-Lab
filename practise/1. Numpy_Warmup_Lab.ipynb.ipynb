## Why NumPy?

You can do numerical calculations using pure Python. In the beginning, you might think Python is fast but once your data gets large, you'll start to notice slow downs.

One of the main reasons you use NumPy is because it's fast. Behind the scenes, the code has been optimized to run using C. Which is another programming language, which can do things much faster than Python.

The benefit of this being behind the scenes is you don't need to know any C to take advantage of it. You can write your numerical computations in Python using NumPy and get the added speed benefits.

If your curious as to what causes this speed benefit, it's a process called vectorization. [Vectorization](https://en.wikipedia.org/wiki/Vectorization) aims to do calculations by avoiding loops as loops can create potential bottlenecks.

NumPy achieves vectorization through a process called [broadcasting](https://docs.scipy.org/doc/numpy/user/basics.broadcasting.html#module-numpy.doc.broadcasting).

## Array attributes
* An array is usually a fixed-size container of items of the same type and size.
* The number of dimensions and items in an array is defined by its shape.
* The shape of an array is a tuple of non-negative integers that specify the sizes of each dimension.

* In NumPy, dimensions are called axes. This means that if you have a 2D array that looks like this:

[[0., 0., 0.],
 [1., 1., 1.]]
* Your array has 2 axes. The first axis has a length of 2 and the second axis has a length of 3.

The more important attributes of an ndarray object are:

*  ndarray.ndim: the number of axes (dimensions) of the array.
* ndarray.shape: the dimensions of the array. This is a tuple of integers indicating the size of the array in each dimension. For a matrix with n rows and m columns, shape will be (n,m). The length of the shape tuple is therefore the number of axes, ndim.
* ndarray.size: the total number of elements of the array. This is equal to the product of the elements of shape.
* ndarray.dtype: an object describing the type of the elements in the array. One can create or specify dtype’s using standard Python types. Additionally NumPy provides types of its own. numpy.int32, numpy.int16, and numpy.float64 are some examples.

## Matrix operations:
Unlike in many matrix languages, the product operator * operates elementwise in NumPy arrays. The matrix product can be performed using the @ operator (in python >=3.5) or the dot function

```python
A = np.array([[1, 1],

              [0, 1]])

B = np.array([[2, 0],

              [3, 4]])

print(A * B)     # elementwise product

print(A @ B)     # matrix product

print(A.dot(B))  # another matrix product
```

## Comparison Operators
Comparison operators evaluate individual elements against a scalar value or match them against elements of another array.

comparison.png

```python
arr1 = np.array([10, 20, 30])
arr2 = np.array([15, 20, 25])

# Comparison with a scalar
print(arr1 > 15)        # Output: [False  True  True]

# Comparison between two arrays
print(arr1 == arr2)     # Output: [False  True False]
```

When combining multiple conditions on NumPy arrays, you cannot use standard Python keywords (and, or, not). You must use NumPy’s universal logical functions or bitwise operators.

### Practical Applications
### 1. Boolean Indexing (Filtering)
You can use a Boolean array as a mask to extract the actual elements that meet your conditions.

```python
arr = np.array([5, 12, 18, 25])
mask = arr > 15

print(arr[mask])
```

### 2. Conditional Replacements (np.where)
The np.where(condition, value_if_true, value_if_false) function lets you modify elements matching a specific rule

```python
arr = np.array([1, 2, 3, 4])
# Replace odd numbers with -1
new_arr = np.where(arr % 2 != 0, -1, arr)

print(new_arr)
```

### 3. Array-Wide Truth Checks
If you need a single True or False for the entire array instead of an element-wise mask, use np.any() or np.all().
* np.any(): Returns True if at least one element meets the condition.
* np.all(): Returns True only if every element meets the condition.

```python
arr = np.array([1, 2, 3])

print(np.any(arr > 2))
print(np.all(arr > 2))
```

# 7. Array Stacking and Concatenation
They are operations that can be used to merge / combine multiple arrays into a unique final array.

image.png

## Concatenation
Join a sequence of arrays along an existing axis.

* Can only be performed on arrays having the same number of axis
* All the axis must have the same number of elements except the axis on which the concatenation is performed.
* If we start from two arrays of shape [A,B,C,N1] and [A,B,C,N2], and we perform concatenation on the last axis, we obtain a new array of shape [A,B,C,N1+N2] whhere elements in the last axis are concatenated.

```python
arr1 = np.arange(1, 4)
arr2 = np.arange(4, 7)
print(arr1.shape, arr2.shape)
print(arr1)
print(arr2)
```

```python
# Concatenation

# [3] + [3] -> [6]
axis_0 = np.concatenate([arr1, arr2], axis=0)
axis_0
```

```python
arr1 = np.arange(1, 7).reshape(2,3)
arr2 = np.arange(7, 13).reshape(2,3)
print(arr1.shape, arr2.shape)
print(arr1)
print(arr2)
```

```python
# [2,3] + [2,3] -> [4,3]
axis_0 = np.concatenate([arr1, arr2], axis=0)
print(axis_0.shape)
print(axis_0)

# [2,3] + [2,3] -> [2,6]
axis_1 = np.concatenate([arr1, arr2], axis=1)
print(axis_1.shape)
print(axis_1)
```

## Stacking
Join a sequence of arrays along a new axis.

* Can only be applied to arrays having the same number of dimensions and the same number of elements in each dimension.
* If we start from two arrays of shape [A,B,C,D] and [A,B,C,D] and we perform stacking on the last axis, we obtain an array with a new axis in the last position (with number of elements equal to the number of arrays merged): [A,B,C,D,2]

```python
arr1 = np.arange(1, 4)
arr2 = np.arange(4, 7)
print(arr1.shape, arr2.shape)
print(arr1)
print(arr2)
```

```python
# Stacking

# on axis 0
axis_0 = np.stack([arr1, arr2], axis=0)
print(axis_0.shape)
print(axis_0)

# on last axis
axis_1 = np.stack([arr1, arr2], axis=1)
print(axis_1.shape)
print(axis_1)
```
